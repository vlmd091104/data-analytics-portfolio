argument function

```python
num_tuple = (1,2,3,4,5,6)
def num_args(*num):
  print(num[0]*num[1])
```

```python
num_args(*num_tuple)
```

```python
def num_power(num,power):
  print(f'{num} to the power of {power} is {num**power}')
```

```python
num_power(power = 2,num = 3)
```

keyword argument function

```python
def num_kwarg(**num):
  print("My number is: " + num['integer'] + " My other number is: " + num["integer2"])
```

```python
num_kwarg(integer = '12', integer2 = '23')
```
