# Sets

this work like hash set

```python
daily_pints = {1,2,3}
```

```python
type(daily_pints)
```

```python
print(daily_pints)
```

```python
daily_pints_log = {1,2,31,2,3,4,1,2,5,6,3,2}
print(daily_pints_log)
```

```python
personB_daily_pints_log = {2,31,2,5,6,3,2,7,8,9}
```

```python
print(daily_pints_log | personB_daily_pints_log)
```

```python
print(daily_pints_log & personB_daily_pints_log)
```

```python
print(daily_pints_log - personB_daily_pints_log)
```

```python
print(daily_pints_log ^ personB_daily_pints_log)
```

# Dictionaries

```python
dict_cream = {"name": "Minh Duc", "weekly intake": 2, "favorite ice creams": ["strawbery", "mango"]}
```

```python
type(dict_cream)
```

```python
print(dict_cream)
```

```python
dict_cream.values()
```

```python
dict_cream.keys()
```

```python
dict_cream.items()
```

```python
dict_cream["name"]
```

```python
dict_cream["weekly intake"] = 3
```

```python
dict_cream['weekly intake']
```

```python
dict_cream.update({"name": "Minh Duc", "age": 22, "favorite phone": ["Poco f8 pro max", "poco f9 ultra"]})
print(dict_cream)
```

```python
del dict_cream["age"]
print(dict_cream)
```
