```python
import os, shutil
```

```python
path = r"C:/Users/vulem/Downloads/"
```

```python
file_name = os.listdir(path)
```

```python
folder_name = ["csv_file", "text_file", "image_file", "pdf_file"]
for i in range (0,4):
    if not os.path.exists(path + folder_name[i]):
        os.makedirs(path + folder_name[i])
```

```python
for file in file_name:
    if ".csv" in file and not os.path.exists(path + "csv_file/" + file):
        shutil.move(path + file,path + "csv_file/" + file)
    elif ".txt" in file and not os.path.exists(path + "text_file/" + file):
        shutil.move(path + file,path + "text_file/" + file)
    elif ".png" in file and not os.path.exists(path + "image_file/" + file):
        shutil.move(path + file,path + "image_file/" + file)
    elif ".pdf" in file and not os.path.exists(path + "pdf_file/" + file):
        shutil.move(path + file,path + "pdf_file/" + file)
```
